```python
import torch
from torch.utils.data import DataLoader
import os
import sys

# Add "scripts" directory to the Python path
module_path = os.path.abspath(os.path.join(os.path.join("..")))
if module_path not in sys.path:
    sys.path.append(module_path)

from scripts.dataset import BCIDataset

PROCESSED_DATA_DIR = "./../data/processed/"
BATCH_SIZE = 32
```

```python
def test_dataloader(dataloader):
    # Verify a single batch
    X_batch, y_batch = next(iter(dataloader))

    # Object inspection
    print("X_batch shape:", X_batch.shape)
    print("y_batch shape:", y_batch.shape)
    print("X_batch type:", type(X_batch))
    print("y_batch type:", type(y_batch))

    expected_x_shape = [BATCH_SIZE, 1, 22, X_batch.shape[3]] # Assuming 22 EEG channels
    assert list(X_batch.shape) == expected_x_shape, "X_batch shape is incorrect"
    assert list(y_batch.shape) == [BATCH_SIZE], "y_batch shape is incorrect"
    assert X_batch.dtype == torch.float32, "X_batch dtype is incorrect"
    assert y_batch.dtype == torch.int32, "y_batch dtype is incorrect"

    print("All tests passed.")

    print("Example X_batch:", X_batch[0])
    print("Example y_batch:", y_batch[0])
```

# Training Dataset

```python
dataset = BCIDataset(PROCESSED_DATA_DIR, dataset_type = "T")
dataloader = DataLoader(
    dataset,
    batch_size = BATCH_SIZE,
    shuffle = True
)
test_dataloader(dataloader)
```

# Evaluation Dataset

```python
dataset = BCIDataset(PROCESSED_DATA_DIR, dataset_type = "E")
dataloader = DataLoader(
    dataset,
    batch_size = BATCH_SIZE,
    shuffle = True
)
test_dataloader(dataloader)
```
